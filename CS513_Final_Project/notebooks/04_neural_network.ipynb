## Train and test on the three feature steps

```python
from sklearn.neural_network import MLPClassifier

results, fitted = evaluate.run_feature_steps(
    "Neural network",
    MLPClassifier(early_stopping=True, max_iter=300, random_state=config.SEED),
    {"hidden_layer_sizes": [(32,), (64,), (64, 32)], "alpha": [1e-4, 1e-2]},
    kind="scaled")
results.round(3)
```

## Training curve

```python
model = fitted["3_plus_violations"].best_estimator_.named_steps["model"]
fig, ax = plots.new_axes("Neural network: accuracy on the held-out slice during training",
                         xlabel="Epoch", ylabel="Validation accuracy")
ax.plot(range(1, len(model.validation_scores_) + 1), model.validation_scores_,
        color=plots.SERIES[3], linewidth=2)
plt.show()
print("Layers:", model.hidden_layer_sizes, "  Epochs run:", model.n_iter_)
```
